```python
#Pydantic lets you define the shape and rules of your data using Python classes.
# Python dictionary → unstructured  Pydantic model → structured + validated
from pydantic import BaseModel,Field
from typing import List, Annotated
```

```python
class User(BaseModel):
    name: str
    email:str
    age: int

    def display(self):
        print(f"Name: {self.name} \nEmail: {self.email} \nAge: {self.age}\n")

user1 = User(name = "Ritu Singh", email = "ritu23791@gmail.com", age = 21)
user1.display()
user2 = User(name = "Arunima", email = "arunima34895@gmail.com", age="45")
user2.display() # Age type conversion is performed
# user3 = User(name = "Arunima", email = "arunima34895@gmail.com", age="Age is 45")
# user3  # this gives error because here type conversion is not done properly

```

```python
## In pydantic these variable are instance variable not class variable 
## and also along with type of data it store you can also define field value
```

```python
class User(BaseModel):
    name:str = Field(min_length = 0 , max_length = 20, default="N/A")
    email:str = Field(min_length = 6 , max_length = 30)
    age:int = Field(ge=0 , le=100)

    def display(self):
            print(f"Name: {self.name} \nEmail: {self.email} \nAge: {self.age}\n")
    
```

```python
user1 = User(email="parul93752@gamil.com", age=23)
user1.display()

user2 = User(name = "Eti Singh", email="etisingh8934@gamil.com", age=25)
user2.display()

# user3 = User(name = "Mahendra Pratap Singh", email="mahendrapratapsingh8934@gamil.com", age=122)
# user3.display() # error because age value is greater than 100 and also name length is exceeding 20
```

```python
class Product(BaseModel):
    id:str = Field(min_length = 6 , max_length = 12)
    name:str = Field(max_length = 30, default = "N/A")
    stock:int = Field(ge=0, lt=2000)
    price:int = Field(gt=0)
    description:str = Field(max_length = 100, default = "N/A")

    def displayProductDetails(self):
        print(f"Id: {self.id} \nName: {self.name} \nPrice: {self.price} \nStock: {self.stock} \nDescription: {self.description}")

product1 = Product(
    id="nsbbj121542", 
    name="Nike Shoes", 
    stock=40,
    price = 1200
)
product2 = Product(
    id="sdnd24342", 
    name = "iPhone 17", 
    stock = 100, 
    price = 120000,
    description = "Apple phone latest version with best camera quality"
)
product3 = Product(
    id="mbp98231",
    name="MacBook Pro 16",
    stock=25,
    price = 200000,
    description="M3 Max chip with liquid retina XDR display",
)
product4 = Product(
    id="snx88219",
    name="Sony WH-1000XM5 Headphones",
    stock=60,
    price=5000,
    description="Industry-leading noise-canceling wireless headphones",
)

product1.displayProductDetails()
```

```python
productList = Annotated[List[Product], Field(default = [])]
```

```python
class Cart(BaseModel):
    id:str = Field(min_length = 5, max_length = 13)
    items:productList = Field(default = [])
    comment:str = Field(default = "This is my Cart")

    def Add_To_Cart(self,item):
        self.items.append(item)
        print("Item Added in the cart successfully")

    def remove_item(self,removed_item):
        self.items = [item for item in self.items if item.id !=removed_item.id ]
        print("Item removed from the cart successfully")
        # Another Way
        # self.items = list(filter(lambda x: x.id!=removed_item.id,self.items))

    def clear_cart(self):
        self.items.clear()
        print("All the items from the Cart are removes Successfully")
```

```python
class Order(BaseModel):
    order_id:str = Field(min_length=5 , max_length=12)
    item:Product
    bill:int = Field(default=0) 

    def place_order(self):
        self.bill=self.item.price
        print("Order placed successfully",self)
```

```python
cart = Cart(id="sfjb2479")
cart.Add_To_Cart(product1)
cart.Add_To_Cart(product3)
cart.Add_To_Cart(product4)
print(len(cart.items))
cart.remove_item(product3)
print(len(cart.items))

```
