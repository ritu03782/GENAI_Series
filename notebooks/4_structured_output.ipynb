```python
from pydantic import BaseModel, Field ## helps to create class in python
from langchain_core.messages import HumanMessage
from typing import List
```

```python
class ResponseStructure(BaseModel):
    name :str = Field(description = "Complete Name")
    email:str = Field(description = "Email Address")
    age:int = Field(description = "Age")

structured_llm = llm.with_structured_output(ResponseStructure)
res = structured_llm.invoke(f"Please give me name ,email and age only from the given text {text}")
res.model_dump()
```

```python
class Movies(BaseModel):
    title:str = Field(description="Movie Title")
    year:int = Field(description="Movie Release Year")
```

```python
movie_llm = llm.with_structured_output(Movies)
query = input("User: ")
res = movie_llm.invoke([HumanMessage(content=query)])
res.model_dump()
```

```python
class AllMovies(BaseModel):
    movies:List[Movies] = Field(description="List of movies")

all_movie_llm = llm.with_structured_output(AllMovies)
query = input("User: ")
res = all_movie_llm.invoke([HumanMessage(content=query)])
res.model_dump()
```
