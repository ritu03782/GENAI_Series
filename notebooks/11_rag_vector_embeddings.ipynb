```python
from dotenv import load_dotenv
load_dotenv()
```

```python
documents = [
    "Machine learning is a subset of artificial intelligence",
    "Deep learning uses neural networks with multiple layers",
    "Python is a programming language popular for data science",
    "India won the cricket world cup in 2011",
    "AI models can understand and generate human language"
]
```

```python
from langchain_huggingface import HuggingFaceEmbeddings

embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)
```

```python
vector = embeddings.embed_documents(documents)
len(vector)
## number of documents

len(vector[0])
## number of dimensions in each embedding
print(vector)
```

```python
## these vector database helps in similarity search which is not possible in case of SQL and MongoDB because of exact text search
```

##### Store Vector Embeddings in Vector DB

```python
from langchain_chroma import Chroma
```

```python
vector_store = Chroma.from_texts(
    texts= documents,
    embedding=embeddings,
    persist_directory="./vector_db"
)
```

```python
query = "Machine Learnings"
result = vector_store.similarity_search(query)
result  ### give results in descending order of their similarity
```

```python
query1 = "Python"
result1 = vector_store.similarity_search(query, k=2)
result1 ## k=2 so only 2 higher priority documents will be printed
```
