```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_groq import ChatGroq
from langchain.agents import create_agent
from langchain_core.tools import tool
llm = ChatGroq(model="openai/gpt-oss-20b")
```

```python
@tool # decorator to declare tool
def add_numbers(a:int , b:int):
    """
        it will return the sum of two numbers

        Args:
            a : Number One
            b : Number Two
    """  # descriptive string telling the task performed by tool
    return a + b

@tool
def multiply_numbers(a:int , b:int):
    """
        it will return the product of two numbers

        Args:
            a : Number One
            b : Number Two
    """
    return a * b
```

```python
add_numbers.invoke({"a":12, "b":3})
```

```python
agent = create_agent(
    model=llm,
    tools=[add_numbers, multiply_numbers],
    system_prompt="You are a math teacher, and always use tool for calculation."
)
```

```python
response = agent.invoke({"messages":[{"role":"user","content":"What is 2 + 3 ?"}]})
```

```python
for res in response['messages']:
    print(res)
    print("\n")
```
