```python
from langchain_ollama import ChatOllama
llm = ChatOllama(model = "qwen2.5:1.5b")
```

```python
res = llm.invoke("Hello, My name is Ankita.")
res.content
```

```python
res = llm.invoke("What is my name ?")
res.content
```

```python
prompts = [
    {"role":"user","content":"Hello, My name is Ankita."},
    {"role":"ai","content":"Hello Ankita! It's nice to meet you. How can I help you today?"},
    {"role":"user","content":"What is my name ?"},
]
res = llm.invoke(prompts)
res.content
```

# Manage History of conversation

```python
history = []
while True:
    query = input("USER: ")

    if query.lower() in ["quit","bye","exit"]:
        print("GoodBye 👋")
        break

    history.append({"role":"user","content":query})
    print("USER: ",query)

    res = llm.invoke(history)
    history.append({"role":"ai","content":res.content})
    print("AI: ",res.content,"\n")
```
