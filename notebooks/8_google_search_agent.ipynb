```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_community.utilities import GoogleSerperAPIWrapper
from langchain_groq import ChatGroq
from langchain.agents import create_agent
```

```python
llm = ChatGroq(model="openai/gpt-oss-20b")
search = GoogleSerperAPIWrapper() # Serper Search object
```

```python
agent = create_agent(
    model=llm,
    tools=[search.run], # search.run is tool of serper search object
    system_prompt="You are an agent and can search for any question on google."
)
```

```python
query = "Which cricket team won the Woman's World Cup 2025 ?"
```

```python
response = agent.invoke({"messages":[{"role":"user","content":query}]})
```
