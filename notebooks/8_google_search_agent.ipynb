```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_community.utilities import GoogleSerperAPIWrapper
from langchain_groq import ChatGroq
from langchain.agents import create_agent
from langgraph.checkpoint.memory import MemorySaver
```

```python
llm = ChatGroq(model="openai/gpt-oss-20b")
search = GoogleSerperAPIWrapper() # Serper Search object
memory = MemorySaver()
```

```python
agent = create_agent(
    model=llm,
    tools=[search.run], # search.run is tool of serper search object
    system_prompt="You are an agent and can search for any question on google.",
    checkpointer = memory
)
```

```python
query = "What were their score ?"
```

```python
response = agent.invoke(
               {"messages":[{"role":"user","content":query}]},
               {"configurable":{"thread_id":"asd123"}}
           )
```
