```python
from langchain_community.document_loaders import PyPDFLoader , TextLoader
```

```python
loader = TextLoader("../data/speech.txt")
pdfloader = PyPDFLoader("../data/GenAI_Interview_Notes.pdf")

textContent = loader.load()
pdfContent = pdfloader.load()
```

```python
splitter = RecursiveCharacterTextSplitter(chunk_size = 1000 , chunk_overlap = 100)
```

```python
text_chunks = splitter.split_documents(textContent)
for textchunk in text_chunks:
    print(textchunk.metadata)
    print(textchunk.page_content)
```

```python
pdf_chunks = splitter.split_documents(pdfContent)
print(pdf_chunks[0].page_content)
```
