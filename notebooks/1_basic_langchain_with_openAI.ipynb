```python
# Here we are using Langchain model - google gemini
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI
```

```python
load_dotenv()
```

```python

llm = ChatGoogleGenerativeAI(model="gemini-3.8-flash")
res = llm.invoke("Which is pm of India?")
print(res.content[0]["text"])
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
llm = ChatGoogleGenerativeAI(model="gemini-3.8-flash")
# Static Prompt
prompt = [
    ("system","Act as python developer"), # first argument is type 'system' ->means how you want your AI to act as python developer
    ("user","Give code to sort an array"), # 'user' ->means query asked by user
      # 'ai' -> means content entered is itself given by ai or refer previous content given by ai
]
res = llm.invoke(prompt)
print(res.content[0]["text"])
```
