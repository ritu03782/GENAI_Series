### Text document loader

```python
from langchain_community.document_loaders import TextLoader
loader = TextLoader("../data/speech.txt")
texts = loader.load()
print(texts[0].page_content)
```

### PDF document loader

```python
from langchain_community.document_loaders import PyPDFLoader
loader = PyPDFLoader("../data/GenAI_Interview_Notes.pdf")
docs = loader.load()
len(docs) ### 26 pages of pdf become 26 values of list named docs
```

```python
oneDoc = docs[0]
print(oneDoc.page_content)
```

##### CSV File Loader

```python
from langchain_community.document_loaders import CSVLoader
loader = CSVLoader("../data/employee_records.csv")
docs = loader.load()
len(docs)
print(docs[0].page_content)
```

##### Web Page Loader

```python
from langchain_community.document_loaders import WebBaseLoader
loader = WebBaseLoader(web_path="https://techsimplus.com/programs")
docs = loader.load()
print(docs[0].page_content)
```

##### Wikipedia Loader

```python
from langchain_community.document_loaders import WikipediaLoader
loader = WikipediaLoader(query="Taj Mahal", load_max_docs=2)
docs = loader.load()
print(docs[0].page_content)
```
